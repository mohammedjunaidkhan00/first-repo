### **Count the total no. of obj +nt in web page (links, img, buttons, inputs)**

```python
from selenium import webdriver
from selenium.webdriver.common.by import By
import time
```

```python
driver = webdriver.Chrome()
driver.get("https://fatafatbooks.com/")
time.sleep(5)
driver.minimize_window()
```

```python
links = len(driver.find_elements(By.TAG_NAME, "a"))
imgs = len(driver.find_elements(By.TAG_NAME, "img"))
inputs = len(driver.find_elements(By.TAG_NAME, "input"))
buttons = len(driver.find_elements(By.TAG_NAME, "button"))
total = links + imgs + inputs + buttons
```

```python
print(f"Total number of Links in web page: {links}")
print(f"Total number of Images in web page: {imgs}")
print(f"Total number of Inputs in web page: {inputs}")
print(f"Total number of Buttons in web page: {buttons}")
print(f"Total number of Element in web page: {total}")
```
