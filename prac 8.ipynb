### **Perfform Operation on Dropdown**

```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import Select
import time
```

```python
driver = webdriver.Chrome()
driver.get("https://qaplayground.com/practice/dropdowns")
driver.minimize_window()
time.sleep(55)
```

```python
elements = driver.find_element(By.ID, "fruitSelect")
dropdown = Select(elements)
options = dropdown.options
```

```python
print("All Options are:")
for i, option in enumerate(options):
    print(f"Option {i+1} is: {option.text}")
```

```python
print(f"Total number of Element: {len(options)}")
```

```python
default_val = dropdown.first_selected_option.text
print(f"Default selected value is: {default_val}")
```
