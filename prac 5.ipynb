### **Web Element Interaction**

```python
from selenium import webdriver
from selenium.webdriver.support.ui import Select
from selenium.webdriver.common.by import By
import time
```

```python
driver = webdriver.Chrome()
driver.get("https://www.selenium.dev/selenium/web/web-form.html")
driver.minimize_window()
time.sleep(2)
```

```python
driver.find_element(By.NAME,  "my-text").send_keys("Junaid")
driver.find_element(By.NAME, "my-password").send_keys("pass555")
driver.find_element(By.NAME, "my-textarea").send_keys("Hello Wrold")
```

```python
dripdown = Select(driver.find_element(By.NAME, "my-select"))
dripdown.select_by_visible_text("One")
```

```python
driver.find_element(By.NAME, "my-datalist").send_keys("Mumbai")

checkbox = driver.find_element(By.ID, "my-check-2")
if not checkbox.is_selected:
    checkbox.click()

radiobtn = driver.find_element(By.ID, "my-radio-2")
if not radiobtn.is_selected:
    radiobtn.click()
```
