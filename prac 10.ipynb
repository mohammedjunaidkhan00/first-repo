### **Web Table handling and Sorting Validation**

```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import Select
import time
```

```python
driver = webdriver.Chrome()
driver.get("file:///D:/clg/EX/SQA/10 table.html")
driver.minimize_window()
time.sleep(5)
```

```python
print("Table Data:")
rows = driver.find_elements(By.CSS_SELECTOR, "table tr") # what is CSS_SELECTOR
for row in rows[1: ]:
    print(row.text)
```

#### **Get Name**

```python
driver.find_element(By.XPATH , "//th[text() = 'Name']").click() 
```

```python
rows = driver.find_elements(By.CSS_SELECTOR, "table tr") # what is CSS_SELECTOR
names = []
```

```python
for row in rows[1:]:
    cell = row.find_elements(By.TAG_NAME, "td")
    names.append(cell[0].text)
```

```python
print(f"Sorted: {names}")
print(f"Sorting Correct: {names == sorted(names)}")
```

#### **Applying Filter**

```python
search = driver.find_element(By.ID, "search")
search.send_keys("John")

rows = driver.find_elements(By.CSS_SELECTOR, "table tr") # what is CSS_SELECTOR

for row in rows[1:]:
    if row.is_displayed():
        print(row.text)
```
